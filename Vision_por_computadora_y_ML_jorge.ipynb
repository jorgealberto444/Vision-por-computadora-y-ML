```python
# Importar la clase Sequential
# Sequential permite crear redes neuronales capa por capa en orden secuencial
from tensorflow.keras.models import Sequential

# Importar la clase Dense
# Dense representa una capa totalmente conectada (fully connected layer)
from tensorflow.keras.layers import Dense


# Crear el modelo de red neuronal usando la estructura secuencial
# Esto significa que las capas se ejecutan una después de otra
Model = Sequential([

    # Primera capa (capa oculta)
    # Dense(16) → crea 16 neuronas en esta capa
    # activation='relu' → función de activación ReLU
    # input_shape=(10,) → la red recibe 10 valores de entrada
    #
    # Esto significa que cada muestra de entrada tiene 10 características
    Dense(16, activation='relu', input_shape=(10,)),


    # Segunda capa oculta
    # Dense(8) → crea 8 neuronas
    # activation='relu' → introduce no linealidad en el modelo
    #
    # Esta capa procesa la información generada por la capa anterior
    Dense(8, activation='relu'),


    # Capa de salida
    # Dense(1) → solo una neurona de salida
    # activation='sigmoid' → produce un valor entre 0 y 1
    #
    # Se utiliza principalmente para:
    # - clasificación binaria
    # - probabilidad de una clase
    Dense(1, activation='sigmoid')

])


# Mostrar el resumen del modelo
# summary() muestra:
# - número de capas
# - número de neuronas por capa
# - número de parámetros (pesos y bias)
# - forma de entrada y salida
Model.summary()
```

```python
# Importa la función para generar un conjunto de datos de clasificación sintético
from sklearn.datasets import make_classification

# Importa el modelo de red neuronal multicapa (Perceptrón Multicapa)
from sklearn.neural_network import MLPClassifier


# Genera un conjunto de datos artificial para clasificación
# n_samples=200 → crea 200 muestras (filas)
# n_features=4 → cada muestra tendrá 4 características (columnas)
# X → matriz de características (variables de entrada)
# y → vector de etiquetas (clases objetivo)
X, y = make_classification(n_samples=200, n_features=4)


# Crea el modelo de red neuronal
# hidden_layer_sizes=(10,10) → dos capas ocultas con 10 neuronas cada una
# max_iter=500 → máximo de 500 iteraciones durante el entrenamiento
clf = MLPClassifier(hidden_layer_sizes=(10,10), max_iter=500)


# Entrena el modelo usando los datos generados
# X → datos de entrada
# y → etiquetas reales
clf.fit(X, y)


# Evalúa el modelo usando los mismos datos de entrenamiento
# score() devuelve la exactitud (accuracy), es decir:
# (número de predicciones correctas / total de muestras)
print("precision:", clf.score(X, y))
```

```python
#EJERCICIO 1
import numpy as np # Importa la librería NumPy para el manejo de arreglos multidimensionales (matrices)
from scipy.signal import convolve2d # Importa la función específica para realizar la operación matemática de convolución en 2D

# Definimos una matriz de 4x4 que representa una imagen digital de baja resolución
# Los valores '10' representan un fondo oscuro y los '50' un objeto o área clara en el centro
imagen = np.array([
    [10, 10, 10, 10],
    [10, 50, 50, 10],
    [10, 50, 50, 10],
    [10, 10, 10, 10]
])

# Definimos el kernel o filtro de 3x3 que se deslizará sobre la imagen
# Este diseño específico (8 en el centro rodeado de -1) es un filtro Laplaciano
# Su función es calcular la diferencia de intensidad entre un píxel y sus vecinos para hallar bordes
kernel = np.array([
    [-1, -1, -1],
    [-1,  8, -1],
    [-1, -1, -1]
])

# Ejecuta la operación de convolución entre la matriz de la imagen y la del filtro
# mode='valid': El filtro solo se aplica donde se superpone completamente con la imagen
# Esto significa que no se usa relleno (padding), por lo que la salida será de 2x2 en lugar de 4x4
resultado = convolve2d(imagen, kernel, mode='valid')

# Muestra en la terminal la matriz resultante con las características extraídas
print(resultado)
```

```python

from tensorflow.keras.models import Sequential # Importa la estructura para modelos lineales donde las capas se apilan en secuencia
from tensorflow.keras.layers import Conv2D # Importa la capa de convolución bidimensional para procesar datos espaciales (imágenes)

# Inicializa el contenedor del modelo donde añadiremos los componentes de la arquitectura
model = Sequential([

    # Define la primera capa de extracción de características
    Conv2D(
        filters=8,              # Especifica que la capa generará 8 mapas de características distintos mediante 8 kernels
        kernel_size=(3, 3),     # Define la ventana de escaneo de 3x3 que buscará patrones locales en la imagen
        activation='relu',      # Aplica la Rectified Linear Unit para descartar valores negativos y permitir el aprendizaje de patrones complejos
        input_shape=(28, 28, 1) # Establece las dimensiones de entrada: 28px de alto, 28px de ancho y 1 canal de color (monocromático)
    )
])

# Genera un reporte en consola con la arquitectura, dimensiones de salida y el conteo de pesos (weights) y sesgos (biases)
model.summary()
```

```python
#Ejercicio 2
import tensorflow as tf # Importa el ecosistema principal para el desarrollo de modelos de aprendizaje profundo

# Descarga y divide automáticamente el conjunto de datos MNIST en lotes de entrenamiento (60k) y evaluación (10k)
# x representa los datos de entrada (matrices de píxeles) y y las etiquetas (el número real del 0 al 9)
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Imprime las dimensiones actuales del tensor de entrenamiento para verificar la cantidad y tamaño de las muestras
# El resultado (60000, 28, 28) indica que hay 60 mil matrices de 28 filas por 28 columnas
print(x_train.shape)

# Realiza un escalado de características (Feature Scaling) dividiendo cada píxel por su valor máximo (255)
# Esto convierte el rango de intensidad [0, 255] a [0.0, 1.0], facilitando la convergencia del gradiente
x_train = x_train / 255
x_test = x_test / 255

# Redimensiona el tensor para cumplir con el formato (batch, height, width, channels) requerido por capas Conv2D
# El parámetro -1 actúa como un comodín que preserva el número total de imágenes originales (60,000)
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# Imprime la forma final del tensor para confirmar la adición de la cuarta dimensión (el canal de profundidad)
# La salida esperada es (60000, 28, 28, 1)
print(x_train.shape)
```

```python
import matplotlib.pyplot as plt

# Mostramos la primera imagen del conjunto de entrenamiento
# x_train[0] es una imagen de 28x28 píxeles (con canal si ya hiciste reshape)
plt.imshow(x_train[0], cmap='gray')

# Añadimos el título con su etiqueta real (el número que representa la imagen)
plt.title(f"Etiqueta: {y_train[0]}")

# Ocultamos los ejes para una visualización más limpia
plt.axis("off")

# Mostramos la imagen en pantalla
plt.show()
```

```python
from tensorflow.keras import Sequential # Importa el framework para construir modelos mediante la acumulación lineal de capas
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout # Importa los componentes esenciales: filtros, submuestreo, aplanado, capas densas y regularizadores

# Inicializa la estructura del modelo para el procesamiento secuencial de tensores
model = Sequential([

    # Primera etapa de extracción: Aplica 32 kernels de 3x3 para identificar rasgos básicos (bordes, puntos)
    Conv2D(
        32, (3, 3),
        activation='relu',
        input_shape=(28, 28, 1)  # Define la entrada para imágenes MNIST de un solo canal
    ),

    # Reducción de dimensionalidad: Toma el valor máximo en ventanas de 2x2 para reducir el ancho y alto a la mitad
    MaxPooling2D(pool_size=(2, 2)),

    # Segunda etapa de extracción: Incrementa a 64 filtros para detectar patrones más complejos (curvas, formas)
    Conv2D(
        64, (3, 3),
        activation='relu'
    ),

    # Segunda reducción: Optimiza el costo computacional y ayuda a que la red sea invariante a pequeñas traslaciones
    MaxPooling2D(pool_size=(2, 2)),

    # Transformación de datos: Convierte el mapa de características 2D (matriz) en un vector unidimensional (arreglo lineal)
    Flatten(),

    # Capa de inferencia: Capa oculta con 128 neuronas para razonar sobre las características extraídas previamente
    Dense(
        128,
        activation='relu'
    ),

    # Técnica de regularización: Desactiva el 50% de las neuronas al azar para forzar a la red a no depender de pesos específicos
    Dropout(0.5),

    # Capa de decisión final: Genera un vector de 10 posiciones (una por cada dígito del 0 al 9)
    # Softmax garantiza que la suma de las 10 salidas sea igual a 1 (distribución de probabilidad)
    Dense(
        10,
        activation='softmax'
    )
])

# Despliega el resumen de la red: dimensiones de salida de cada capa y el total de pesos a entrenar
model.summary()
```

```python
# Configura el proceso de aprendizaje del modelo mediante la definición de la función de coste y el optimizador
model.compile(
    # Implementa el optimizador 'Adam' (Adaptive Moment Estimation)
    # Este algoritmo ajusta la tasa de aprendizaje de forma individual para cada parámetro, acelerando la convergencia
    optimizer='adam',

    # Define la función de pérdida (Loss Function) para medir la discrepancia entre la predicción y el valor real
    # 'sparse_categorical_crossentropy' es ideal cuando las etiquetas son números enteros (ej. 0, 1, 2...)
    loss='sparse_categorical_crossentropy',

    # Establece los indicadores de rendimiento que se monitorearán durante el proceso
    # 'accuracy' (precisión) reporta la fracción de imágenes clasificadas correctamente sobre el total
    metrics=['accuracy']
)
```

```python
# Evaluamos el modelo usando el conjunto de prueba (datos que NO vio durante entrenamiento)
test_loss, test_acc = model.evaluate(x_test, y_test)

# Mostramos la precisión (accuracy) del modelo en los datos de prueba
print(f"Precisión en el conjunto de prueba:", test_acc)
```

```python
import tensorflow as tf # Importa el motor principal de Deep Learning
from tensorflow.keras import datasets, layers, models # Importa herramientas para conjuntos de datos y construcción de capas
from tensorflow.keras.preprocessing.image import ImageDataGenerator # Librería para aumento de datos (aunque no se usa en este bloque)
import matplotlib.pyplot as plt # Importa la librería para visualización de gráficas y resultados

# Carga el dataset CIFAR-10 y lo descompone en tuplas de entrenamiento y validación
(x_train, y_train), (x_test, y_test) = datasets.cifar10.load_data()

# Transforma los valores de intensidad de los píxeles de tipo entero [0-255] a tipo flotante [0-1]
# Esta normalización es crítica para que el optimizador trabaje en una escala uniforme
x_train = x_train / 255.0
x_test = x_test / 255.0

# Instancia un modelo de tipo Secuencial para definir la arquitectura de la red
cnn_model = models.Sequential([

    # Primera capa convolucional: Extrae 32 mapas de características iniciales (bordes y colores)
    layers.Conv2D(
        32, (3,3),
        activation='relu',
        input_shape=(32,32,3) # Recibe imágenes de 32x32 píxeles con 3 canales (RGB)
    ),

    # Capa de submuestreo: Reduce el tamaño espacial de los mapas de características a la mitad (16x16)
    layers.MaxPooling2D((2,2)),

    # Segunda capa convolucional: Extrae 64 características de mayor jerarquía (formas complejas)
    layers.Conv2D(
        64, (3,3),
        activation='relu'
    ),

    # Segunda capa de submuestreo: Reduce nuevamente las dimensiones espaciales para optimizar parámetros
    layers.MaxPooling2D((2,2)),

    # Capa de aplanado: Desenrolla las matrices 2D en un vector unidimensional para la etapa de clasificación
    layers.Flatten(),

    # Capa densa (MLP): Capa oculta con 128 neuronas para la interpretación lógica de las características
    layers.Dense(
        128,
        activation='relu'
    ),

    # Capa de salida: 10 neuronas que corresponden a las 10 categorías del dataset CIFAR-10
    # Softmax entrega una distribución de probabilidad para cada clase
    layers.Dense(
        10,
        activation='softmax'
    )
])

# Configura el proceso de optimización del modelo
cnn_model.compile(
    optimizer='adam', # Algoritmo de gradiente descendente estocástico adaptativo
    loss='sparse_categorical_crossentropy', # Función de pérdida para etiquetas enteras
    metrics=['accuracy'] # Monitorea el porcentaje de precisión durante el proceso
)

# Imprime un resumen detallado de la cantidad de parámetros por capa y las dimensiones de salida
cnn_model.summary()

# Inicia el proceso de entrenamiento (ajuste de pesos) durante 5 iteraciones completas (épocas)
# history_base almacenará las métricas de error y precisión de cada época para su posterior análisis
history_base = cnn_model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_data=(x_test, y_test) # Evalúa el rendimiento con datos no vistos al final de cada época
)
```

```python
from tensorflow.keras.layers import Dropout # Importa la capa de regularización por exclusión aleatoria de neuronas

# Configura un motor de transformación de imágenes en tiempo real para robustecer el entrenamiento
# Estas técnicas evitan que el modelo se memorice la posición exacta de los objetos
datagen = ImageDataGenerator(
    rotation_range=20,      # Rota aleatoriamente las imágenes hasta 20 grados
    zoom_range=0.2,         # Aplica un zoom aleatorio del 20% hacia adentro o hacia afuera
    horizontal_flip=True,   # Invierte la imagen horizontalmente (útil para objetos con simetría como autos)
    width_shift_range=0.1,  # Desplaza la imagen horizontalmente un 10% de su ancho total
    height_shift_range=0.1  # Desplaza la imagen verticalmente un 10% de su alto total
)

# Calcula las estadísticas necesarias (como la media o desviación si se requieren) sobre el set de entrenamiento
datagen.fit(x_train)

# Definición de la arquitectura mejorada para mayor capacidad de generalización
cnn_mejorada = models.Sequential([

    # Bloque de extracción 1: Identificación de rasgos espaciales iniciales
    layers.Conv2D(
        32, (3,3),
        activation='relu',
        input_shape=(32,32,3)
    ),
    layers.MaxPooling2D((2,2)),

    # Bloque de extracción 2: Filtrado de características de alto nivel
    layers.Conv2D(
        64, (3,3),
        activation='relu'
    ),
    layers.MaxPooling2D((2,2)),

    # Transición: Aplanado del volumen a un vector de características
    layers.Flatten(),

    # Capa de razonamiento: 128 neuronas para clasificar los rasgos extraídos
    layers.Dense(
        128,
        activation='relu'
    ),

    # Técnica de Regularización: Desconecta el 50% de las conexiones de forma aleatoria en cada paso
    # Esto evita el co-adaptamiento de neuronas y reduce drásticamente el sobreajuste
    Dropout(0.5),

    # Capa de salida: Distribución de probabilidad sobre las 10 clases del dataset
    layers.Dense(
        10,
        activation='softmax'
    )
])

# Compilación del modelo con optimización de gradiente adaptativo y función de pérdida para etiquetas enteras
cnn_mejorada.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Ejecuta el entrenamiento utilizando el generador de datos (Data Augmentation)
# datagen.flow genera lotes de imágenes transformadas "al vuelo" en cada época
history_mejorada = cnn_mejorada.fit(
    datagen.flow(x_train, y_train, batch_size=64), # Lotes de 64 imágenes procesadas simultáneamente
    epochs=5,                                      # Número de ciclos de entrenamiento sobre el dataset
    validation_data=(x_test, y_test)               # Validación con datos originales (sin transformaciones) para medir precisión real
)
```

```python
#Ejercicio 5
import tensorflow as tf # Importa el framework de aprendizaje profundo
from tensorflow.keras import datasets, layers, models # Herramientas para gestión de datos y arquitectura
from tensorflow.keras.preprocessing.image import ImageDataGenerator # Clase para la generación de variaciones de imágenes
import matplotlib.pyplot as plt # Librería para visualización de métricas

# Importación y segmentación del dataset CIFAR-10 (60,000 imágenes de 32x32 píxeles)
(x_train, y_train), (x_test, y_test) = datasets.cifar10.load_data()

# Normalización de los valores de píxel al rango [0, 1] mediante punto flotante
x_train = x_train / 255.0
x_test = x_test / 255.0

# Función constructora para asegurar que ambos experimentos utilicen la misma arquitectura inicial
def crear_modelo():
    model = models.Sequential([
        # Extractor 1: Identificación de patrones espaciales básicos con 32 núcleos
        layers.Conv2D(32, (3,3), activation='relu', input_shape=(32,32,3)),
        layers.MaxPooling2D((2,2)), # Reducción de resolución espacial (Submuestreo)

        # Extractor 2: Identificación de rasgos de mayor nivel con 64 núcleos
        layers.Conv2D(64, (3,3), activation='relu'),
        layers.MaxPooling2D((2,2)),

        # Bloque de Clasificación: Aplanado de tensores y capas densas de decisión
        layers.Flatten(),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax') # Salida multiclase probabilística
    ])

    # Configuración del proceso de optimización del modelo
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model

# --- EXPERIMENTO 1: Entrenamiento estándar ---
modelo_normal = crear_modelo()

# Ajuste del modelo utilizando únicamente el dataset original estático
history_normal = modelo_normal.fit(
    x_train, y_train,
    epochs=5,
    validation_data=(x_test, y_test)
)

# --- EXPERIMENTO 2: Entrenamiento con aumento de datos ---

# Configuración del motor de transformaciones aleatorias
# Estas operaciones crean variantes de las imágenes originales en cada paso del entrenamiento

datagen = ImageDataGenerator(
    rotation_range=20,      # Invariancia a la rotación: gira la imagen hasta 20°
    zoom_range=0.2,         # Invariancia a la escala: aplica zoom del 20%
    horizontal_flip=True,   # Invariancia a la reflexión: voltea la imagen lateralmente
    width_shift_range=0.1,  # Invariancia a la traslación: desplaza horizontalmente un 10%
    height_shift_range=0.1  # Invariancia a la traslación: desplaza verticalmente un 10%
)

# Calcula parámetros internos necesarios para las transformaciones basadas en el set de entrenamiento
datagen.fit(x_train)

modelo_aug = crear_modelo()

# Entrenamiento dinámico: datagen.flow suministra imágenes transformadas "al vuelo"
# Esto significa que el modelo rara vez ve exactamente la misma imagen dos veces
history_aug = modelo_aug.fit(
    datagen.flow(x_train, y_train, batch_size=64), # Generador de lotes dinámicos
    epochs=5,
    validation_data=(x_test, y_test)
)
```

```python
### **Creación de arqutiectura CNN - Modelo básico**

from keras import layers
from keras import models

model1 = models.Sequential()

# ETAPA DE EXTRACCIÓN DE CARACTERISTICAS
model1.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)))
model1.add(layers.MaxPooling2D((2, 2)))

model1.add(layers.Conv2D(64, (3, 3), activation='relu'))
model1.add(layers.MaxPooling2D((2, 2)))

model1.add(layers.Conv2D(128, (3, 3), activation='relu'))
model1.add(layers.MaxPooling2D((2, 2)))

model1.add(layers.Conv2D(128, (3, 3), activation='relu'))
model1.add(layers.MaxPooling2D((2, 2)))
# -----------------------------------------------------------

# ETAPA DE CLASIFICACIÓN
model1.add(layers.Flatten())
model1.add(layers.Dropout(0.5))
model1.add(layers.Dense(512, activation='relu'))
model1.add(layers.Dense(1, activation='sigmoid'))

model1.summary()
```

### Definir y entrenar la arquitectura - Etapa de clasificación

```python
from keras import models
from keras import layers, optimizers

model = models.Sequential()
model.add(conv_base)
model.add(layers.Flatten())
model.add(layers.Dense(256, activation='relu'))
model.add(layers.Dropout(0.5))
model.add(layers.Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(optimizer=optimizers.RMSprop(learning_rate=2e-5),
              loss='binary_crossentropy',
              metrics=['acc'])
```

### Data Augmentation

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(rescale=1./255,
                                   rotation_range=40,
                                   width_shift_range=0.2,
                                   height_shift_range=0.2,
                                   shear_range=0.2,
                                   zoom_range=0.2,
                                   horizontal_flip=True,
                                   fill_mode='nearest')

test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(train_dir,
                                                    target_size=(150, 150),
                                                    batch_size=20,
                                                    class_mode='binary')

validation_generator = test_datagen.flow_from_directory(validation_dir,
                                                        target_size=(150, 150),
                                                        batch_size=20,
                                                        class_mode='binary')
```

### **Fitting the model**

```python
history = model.fit(train_generator, steps_per_epoch=100, epochs=30,
                     validation_data=validation_generator, validation_steps=50)
```

```python
import matplotlib.pyplot as plt

acc = history.history['acc']
val_acc = history.history['val_acc']
loss = history.history['loss']
val_loss = history.history['val_loss']

epochs = range(1, len(acc) + 1)

plt.plot(epochs, acc, 'bo', label = 'Training acc')
plt.plot(epochs, val_acc, 'b', label = 'Validation acc')
plt.title('Training and validation accuracy')
plt.legend()

plt.figure()

plt.plot(epochs, loss, 'bo', label = 'Training loss')
plt.plot(epochs, val_loss, 'b', label = 'Validation loss')
plt.title('Training and validation loss')
plt.legend()

plt.show()
```
